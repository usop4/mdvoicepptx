```python

s = """
const DEFAULT_CSV = `# sample
가게부,家計簿
각자,各自
감탄,感嘆
격려,激励
관찰력,観察力
난방,暖房
남녀평등,男女平等
다수,多数
단계,段階
단체,団体
"""

# 各行の先頭文字から子音を取り出し、辞書形式で出力
result = {}
for line in s.splitlines():
    if not line.strip():
        continue
    first_char = line[0]
    initial, _ = get_initial_and_vowel(first_char)
    if initial:
        result.setdefault(initial, []).append(first_char)

print(json.dumps(result, ensure_ascii=False, indent=2))
```
